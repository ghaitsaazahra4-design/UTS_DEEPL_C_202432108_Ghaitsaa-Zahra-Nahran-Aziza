  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>

```python
#202432108_GhaitsaaZahraNahranAziza

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.optim as optim

from torch.utils.data import TensorDataset, DataLoader
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432108_GhaitsaaZahraNahranAziza

# 1. Membaca dataset
train = pd.read_csv("DailyDelhiClimateTrain.csv")
test = pd.read_csv("DailyDelhiClimateTest.csv")

# 2. Mengubah kolom tanggal menjadi format datetime
train["date"] = pd.to_datetime(train["date"])
test["date"] = pd.to_datetime(test["date"])

# 3. Mengurutkan data berdasarkan tanggal
train = train.sort_values("date").reset_index(drop=True)
test = test.sort_values("date").reset_index(drop=True)

# 4. Menghapus tanggal yang tumpang tindih dengan data training
test = test[test["date"] > train["date"].max()].reset_index(drop=True)

# 5. Memilih atribut target yaitu suhu rata-rata
train_temp = train["meantemp"].values.astype(np.float32)
test_temp = test["meantemp"].values.astype(np.float32)

# 6. Membagi data training menjadi 80% data latih dan 20% validasi
split_index = int(len(train_temp) * 0.8)

train_part = train_temp[:split_index]
val_part = train_temp[split_index:]

# 7. Normalisasi data menggunakan nilai minimum dan maksimum data latih
data_min = train_part.min()
data_max = train_part.max()

train_scaled = (train_part - data_min) / (data_max - data_min)
val_scaled = (val_part - data_min) / (data_max - data_min)
test_scaled = (test_temp - data_min) / (data_max - data_min)

# 8. Membuat fungsi sliding window
def create_sequences(data, window_size):
    X = []
    y = []

    for i in range(len(data) - window_size):
        X.append(data[i:i + window_size])
        y.append(data[i + window_size])

    return np.array(X), np.array(y)

# 9. Menentukan panjang urutan data
window_size = 30

# 10. Membuat data latih
X_train, y_train = create_sequences(train_scaled, window_size)

# 11. Membuat data validasi dengan menggunakan 30 hari sebelumnya
val_input = np.concatenate((train_scaled[-window_size:], val_scaled))
X_val, y_val = create_sequences(val_input, window_size)

# 12. Membuat data pengujian menggunakan 30 hari terakhir
# dari data training dan validasi sebagai konteks prediksi

previous_data = np.concatenate((train_scaled, val_scaled))

test_input = np.concatenate((
    previous_data[-window_size:],
    test_scaled
))

X_test, y_test = create_sequences(
    test_input,
    window_size
)

# 13. Mengubah data menjadi PyTorch Tensor
X_train = torch.tensor(X_train, dtype=torch.float32).unsqueeze(-1)
y_train = torch.tensor(y_train, dtype=torch.float32).unsqueeze(-1)

X_val = torch.tensor(X_val, dtype=torch.float32).unsqueeze(-1)
y_val = torch.tensor(y_val, dtype=torch.float32).unsqueeze(-1)

X_test = torch.tensor(X_test, dtype=torch.float32).unsqueeze(-1)
y_test = torch.tensor(y_test, dtype=torch.float32).unsqueeze(-1)

# 14. Membuat DataLoader untuk proses training
batch_size = 32

train_dataset = TensorDataset(X_train, y_train)
train_loader = DataLoader(
    train_dataset,
    batch_size=batch_size,
    shuffle=True
)

# 15. Menampilkan informasi hasil preprocessing
print("Jumlah Data Training:", len(X_train))
print("Jumlah Data Validasi:", len(X_val))
print("Jumlah Data Testing:", len(X_test))

print("Bentuk X_train:", X_train.shape)
print("Bentuk y_train:", y_train.shape)
print("Preprocessing berhasil dilakukan.")
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432108_GhaitsaaZahraNahranAziza

class SequenceModel(nn.Module):
    def __init__(self):
        super(SequenceModel, self).__init__()

        # Layer LSTM
        self.lstm = nn.LSTM(
            input_size=1,
            hidden_size=64,
            num_layers=2,
            batch_first=True,
            dropout=0.2
        )

        # Layer untuk menghasilkan prediksi suhu
        self.fc = nn.Linear(64, 1)

    def forward(self, x):

        # Memproses urutan data menggunakan LSTM
        output, (hidden, cell) = self.lstm(x)

        # Mengambil output pada time step terakhir
        last_output = output[:, -1, :]

        # Menghasilkan prediksi suhu
        prediction = self.fc(last_output)

        return prediction


# Membuat objek model
model = SequenceModel()

# Menampilkan arsitektur model
print(model)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432108_GhaitsaaZahraNahranAziza

# 1. Menentukan hyperparameter
epochs = 50
learning_rate = 0.001
hidden_size = 64

# 2. Menentukan model
model = SequenceModel()

# 3. Menentukan fungsi loss
criterion = nn.MSELoss()

# 4. Menentukan optimizer
optimizer = optim.Adam(
    model.parameters(),
    lr=learning_rate
)

# 5. Menyiapkan riwayat loss
train_losses = []
val_losses = []

# 6. Proses training
for epoch in range(epochs):

    model.train()
    running_loss = 0.0

    for batch_X, batch_y in train_loader:

        # Menghapus gradien sebelumnya
        optimizer.zero_grad()

        # Melakukan prediksi
        outputs = model(batch_X)

        # Menghitung nilai loss
        loss = criterion(outputs, batch_y)

        # Menghitung gradien
        loss.backward()

        # Memperbarui parameter model
        optimizer.step()

        running_loss += loss.item() * batch_X.size(0)

    # Menghitung rata-rata training loss
    epoch_loss = running_loss / len(train_loader.dataset)
    train_losses.append(epoch_loss)

    # Menghitung validation loss
    model.eval()
    val_loss = 0.0

    with torch.no_grad():

        for i in range(len(X_val)):
            val_output = model(X_val[i:i+1])
            loss = criterion(val_output, y_val[i:i+1])
            val_loss += loss.item()

    val_loss = val_loss / len(X_val)
    val_losses.append(val_loss)

    # Menampilkan hasil setiap epoch
    print(
        f"Epoch [{epoch+1}/{epochs}], "
        f"Training Loss: {epoch_loss:.6f}, "
        f"Validation Loss: {val_loss:.6f}"
    )

print("Proses training selesai.")
```

```python
#202432108_GhaitsaaZahraNahranAziza

# Mengambil parameter LSTM hasil training
model.eval()

# Menggunakan satu input aktual dari data training
x_t = X_train[0, 0, 0].item()

# Mengambil bobot dan bias dari layer LSTM pertama
lstm = model.lstm

W_ih = lstm.weight_ih_l0.detach()
W_hh = lstm.weight_hh_l0.detach()
b_ih = lstm.bias_ih_l0.detach()
b_hh = lstm.bias_hh_l0.detach()

# Memilih satu unit hidden state, yaitu unit ke-0
unit = 0

# PyTorch menggunakan urutan gate: Input, Forget, Candidate, Output
i_idx = unit
f_idx = 64 + unit
g_idx = 128 + unit
o_idx = 192 + unit

# Menghitung nilai setiap gate menggunakan bobot aktual
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def calculate_gate(idx, activation):
    z = (
        W_ih[idx, 0].item() * x_t
        + b_ih[idx].item()
        + b_hh[idx].item()
    )
    return activation(z), z

i_t, z_i = calculate_gate(i_idx, sigmoid)
f_t, z_f = calculate_gate(f_idx, sigmoid)
g_t, z_g = calculate_gate(g_idx, np.tanh)
o_t, z_o = calculate_gate(o_idx, sigmoid)

# Kondisi awal hidden state dan cell state adalah nol
c_prev = 0.0
h_prev = 0.0

# Menghitung cell state dan hidden state
c_t = f_t * c_prev + i_t * g_t
h_t = o_t * np.tanh(c_t)

# Menampilkan parameter dan hasil perhitungan
print("PERHITUNGAN MANUAL LSTM BERDASARKAN PARAMETER HASIL TRAINING")
print("Nilai input x_t:", x_t)
print("Unit hidden state:", unit)

print("\nBobot dan bias aktual:")
print("Forget Gate - W:", W_ih[f_idx, 0].item())
print("Forget Gate - b:", (b_ih[f_idx] + b_hh[f_idx]).item())
print("Input Gate - W:", W_ih[i_idx, 0].item())
print("Input Gate - b:", (b_ih[i_idx] + b_hh[i_idx]).item())
print("Candidate - W:", W_ih[g_idx, 0].item())
print("Candidate - b:", (b_ih[g_idx] + b_hh[g_idx]).item())
print("Output Gate - W:", W_ih[o_idx, 0].item())
print("Output Gate - b:", (b_ih[o_idx] + b_hh[o_idx]).item())

print("\nHasil perhitungan:")
print("Forget Gate (f_t):", f_t)
print("Input Gate (i_t):", i_t)
print("Candidate Cell State:", g_t)
print("Cell State (C_t):", c_t)
print("Output Gate (o_t):", o_t)
print("Hidden State (h_t):", h_t)
```

  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>

```python
#202432108_GhaitsaaZahraNahranAziza

# 1. Menampilkan grafik Training Loss dan Validation Loss
plt.figure(figsize=(10, 5))

plt.plot(train_losses, label="Training Loss")
plt.plot(val_losses, label="Validation Loss")

plt.title("Grafik Training Loss dan Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.show()


# 2. Mengubah model ke mode evaluasi
model.eval()

# 3. Melakukan prediksi menggunakan data testing
with torch.no_grad():
    predictions_scaled = model(X_test).numpy().flatten()

# 4. Mengubah hasil prediksi ke skala suhu asli
predictions = (
    predictions_scaled * (data_max - data_min)
) + data_min

actual = (
    y_test.numpy().flatten() * (data_max - data_min)
) + data_min


# 5. Menghitung nilai evaluasi model
rmse = np.sqrt(np.mean((actual - predictions) ** 2))
mae = np.mean(np.abs(actual - predictions))

print("Hasil Evaluasi Model LSTM")
print("RMSE:", round(rmse, 4))
print("MAE :", round(mae, 4))


# 6. Membuat grafik perbandingan suhu aktual dan prediksi
plt.figure(figsize=(12, 5))

plt.plot(
    test["date"],
    actual,
    label="Suhu Aktual",
    marker="o"
)

plt.plot(
    test["date"],
    predictions,
    label="Prediksi LSTM",
    marker="x"
)

plt.title("Perbandingan Suhu Aktual dan Prediksi LSTM")
plt.xlabel("Tanggal")
plt.ylabel("Suhu Rata-rata (°C)")
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
